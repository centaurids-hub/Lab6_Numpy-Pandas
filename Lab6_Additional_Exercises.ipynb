Additional Exercise # 1: Programming Language Growth Analysis

Functionality: Analyze the overall growth trajectory and statistical characteristics of your assigned programming language from 2004 to 2024 using NumPy and Pandas operations. The goal is to determine the month-to-month growth rate, identify stable and changing periods, and compute descriptive statistics that summarize its long-term performance.

Input:

    • Dataset file: programming_languages_popularity.csv

    • Assigned language based on the last three digits of the student ID.

    • Sort all language columns alphabetically and select the corresponding one based on the computed index.

Calculations:

    • Read the dataset and convert the date column into datetime format for time-based analysis.

    • Assign a language based on the computed index derived from the student ID.

    • Extract the time-series data for the assigned language and isolate its popularity values.

    • Compute the month-to-month growth rate as a percentage using the pct_change() function.

    • Calculate rolling statistics such as a moving average and moving standard deviation using a fixed rolling window (e.g., 3-month or 12-month).

    • Classify growth phases using NumPy conditional logic:

        − Growth phase → growth rate > threshold (e.g., +5%)

        − Decline phase → growth rate < threshold (e.g., -5%)

        − Stable phase → between thresholds

    • Generate statistical summaries using Pandas methods such as .mean(), .median(), .std(), and, describe().

    • Determine overall growth percentage from the first to the last record.

    • Count occurrences of each phase (growth, decline, stable) using .value_counts().

Output:

    • Assigned Language: Display the language name assigned based on the computed index.
    
    • DataFrame Output: A table containing the following columns:

        − Month
    
        − Popularity
        
        − Growth_Rate
        
        − Moving_Avg

        − Moving_STD
        
        − Phase

    • Statistical Summary Table: Mean, median, standard deviation, min, max, and quartiles of the popularity values.

    • Growth Phase Counts: Number of periods categorized as growth, decline, or stable.

    • Key Performance Metrics:

        − Initial popularity value

        − Final popularity value

        − Overall growth percentage

        − Number of growth, decline, and stable periods

```python
# ANSWER FOR ADDITIONAL EXERCISE #1: Programming Language Growth Analysis

print("EXERCISE #1: Programming Language Growth Analysis")


# Step 1: Load the dataset
# Note: Make sure the CSV file is in the same folder as this script
try:
    df = pd.read_csv('programming_languages_popularity.csv')
    print("\n Dataset loaded successfully!")
    print(f"Shape: {df.shape[0]} rows × {df.shape[1]} columns")
except FileNotFoundError:
    print("\n Error: File not found. Please download the dataset from Kaggle:")
    print("https://www.kaggle.com/datasets/muhammadroshaanriaz/most-popular-programming-languages-2004-2024")
    # Create sample data for demonstration
    print("\nCreating sample data for demonstration...")
    dates = pd.date_range(start='2004-01-01', periods=240, freq='M')
    df = pd.DataFrame({'Month': dates})
    for lang in languages:
        df[f'{lang} Worldwide (%)'] = np.random.uniform(1, 20, 240)

# Step 2: Convert Month column to datetime format
# This allows us to perform time-based operations
df['Month'] = pd.to_datetime(df['Month'])

# Step 3: Extract the column for our assigned language
# The column name format is usually "Language Worldwide (%)"
lang_column = f'{assigned_language} Worldwide (%)'

# Check if column exists, if not use a generic name
if lang_column not in df.columns:
    # Try to find a matching column
    matching = [col for col in df.columns if assigned_language.lower() in col.lower()]
    if matching:
        lang_column = matching[0]
    else:
        lang_column = df.columns[1]  # Use first data column as fallback

print(f"\n✓ Analyzing column: {lang_column}")

# Step 4: Extract time-series data for the assigned language
# Create a new DataFrame with just Month and the language popularity
lang_data = df[['Month', lang_column]].copy()
lang_data.columns = ['Month', 'Popularity']  # Rename for clarity

# Step 5: Calculate month-to-month growth rate
# pct_change() calculates percentage change from previous row
# multiply by 100 to convert to percentage
lang_data['Growth_Rate'] = lang_data['Popularity'].pct_change() * 100

# Step 6: Calculate rolling (moving) statistics
# rolling(window=12) means we look at 12-month periods
# min_periods=1 allows calculation even with fewer data points at start
lang_data['Moving_Avg'] = lang_data['Popularity'].rolling(window=12, min_periods=1).mean()
lang_data['Moving_STD'] = lang_data['Popularity'].rolling(window=12, min_periods=1).std()

# Step 7: Classify growth phases using NumPy conditional logic
# np.select() works like a series of if-else statements
# Conditions are checked in order, and the first matching condition wins

conditions = [
    lang_data['Growth_Rate'] > 5,    # Growth phase: > 5% increase
    lang_data['Growth_Rate'] < -5,   # Decline phase: > 5% decrease
]
choices = ['Growth', 'Decline']
# Default is 'Stable' if neither condition is met
lang_data['Phase'] = np.select(conditions, choices, default='Stable')

# Step 8: Generate statistical summary
print("\n" + "-" * 40)
print("STATISTICAL SUMMARY")
print("-" * 40)
summary = lang_data['Popularity'].describe()
print(summary)

# Step 9: Calculate key performance metrics
initial_pop = lang_data['Popularity'].iloc[0]  # First value
final_pop = lang_data['Popularity'].iloc[-1]     # Last value
overall_growth = ((final_pop - initial_pop) / initial_pop) * 100

# Count occurrences of each phase
phase_counts = lang_data['Phase'].value_counts()

print("\n" + "-" * 40)
print("KEY PERFORMANCE METRICS")
print("-" * 40)
print(f"Initial Popularity: {initial_pop:.2f}%")
print(f"Final Popularity:   {final_pop:.2f}%")
print(f"Overall Growth:     {overall_growth:.2f}%")
print(f"\nGrowth Phase Counts:")
for phase, count in phase_counts.items():
    print(f"  {phase}: {count} months")

# Step 10: Display sample of the DataFrame
print("\n" + "-" * 40)
print("SAMPLE DATA TABLE (First 10 rows)")
print("-" * 40)
print(lang_data.head(10).to_string())
```

Additional Exercise # 2: Life Cycle Classification of Programming Languages

Functionality: Analyze and classify the lifecycle stage of a programming language based on its popularity trend over time. Using NumPy and Pandas, compute growth metrics, smooth time-series variations, and categorize each language into one of the four stages: Introduction, Growth, Maturity, or Decline.

Input:

    • Dataset file: most-popular-programming-languages-2004-2024.csv (downloaded from Kaggle)

    • Columns used:

        − Month
        
        − One language column (e.g., Python Worldwide (%))

    • Each student will analyze one language, assigned according to their last three digits of the student ID

Calculations:

    • Load and clean the dataset using Pandas (read_csv()), ensuring that the Month column is parsed as a datetime type.

    • Select the assigned language column for individual analysis.

    • Compute growth rate:
    
    • Use pct_change() × 100 to determine month-to-month percentage growth.

    • Smooth the trend:

        − Apply moving average and moving standard deviation using rolling(window=6).mean() and rolling(window=6).std().
    
    • Analyze central tendency:
        
        − Calculate overall mean and standard deviation of the growth rate using NumPy (np.mean() and np.std()).
    
    • Define lifecycle rules:
    
        − Introduction: Growth rate > 0 but below mean

        − Growth: Growth rate > mean
    
        − Maturity: |Growth rate| ≤ 1%

        − Decline: Growth rate < 0 and below (−1 × std)
    
    • Apply lifecycle classification using np.select() or chained conditions.
    
    • Summarize distribution using value_counts() to show how many months fall under each lifecycle phase.
    
    • Identify the dominant lifecycle stage and interpret what it implies about the language’s global trajectory.

Output:
    
    • Lifecycle Classification Table including:

        − Month
    
        − Popularity (%)
    
        − Growth_Rate (%)
    
        − Moving_Avg
    
        − Moving_STD
    
        − Lifecycle_Phase

    • Summary Output:
    
        − Count and percentage per lifecycle phase
        
        − Dominant stage (e.g., “Python is currently in the maturity phase with stable global adoption.”)

Interpretive Insights:

    − Transitions over the years (e.g., Growth → Maturity)
    
    − Indications of possible resurgence or decline

```python
# ANSWER FOR ADDITIONAL EXERCISE #2: Lifecycle Classification of Programming Languages

print("EXERCISE #2: Lifecycle Classification")


# Step 1: Use the same dataset and language from Exercise 1
lifecycle_data = df[['Month', lang_column]].copy()
lifecycle_data.columns = ['Month', 'Popularity']

# Step 2: Calculate growth rate (month-to-month percentage change)
lifecycle_data['Growth_Rate'] = lifecycle_data['Popularity'].pct_change() * 100

# Step 3: Smooth the trend with moving statistics (6-month window)
lifecycle_data['Moving_Avg'] = lifecycle_data['Popularity'].rolling(window=6, min_periods=1).mean()
lifecycle_data['Moving_STD'] = lifecycle_data['Popularity'].rolling(window=6, min_periods=1).std()

# Step 4: Calculate central tendency of growth rate
mean_growth = np.mean(lifecycle_data['Growth_Rate'].dropna())
std_growth = np.std(lifecycle_data['Growth_Rate'].dropna())

print(f"\nGrowth Rate Statistics:")
print(f"  Mean: {mean_growth:.4f}%")
print(f"  Std Dev: {std_growth:.4f}%")

# Step 5: Define lifecycle classification rules
# Introduction: Growth rate > 0 but below mean (early growth, not yet booming)
# Growth: Growth rate > mean (strong positive growth)
# Maturity: |Growth rate| ≤ 1% (stable, minimal change)
# Decline: Growth rate < 0 and below (-1 × std) (significant negative growth)

lifecycle_conditions = [
    (lifecycle_data['Growth_Rate'] > 0) & (lifecycle_data['Growth_Rate'] < mean_growth),
    lifecycle_data['Growth_Rate'] > mean_growth,
    abs(lifecycle_data['Growth_Rate']) <= 1,
    (lifecycle_data['Growth_Rate'] < 0) & (lifecycle_data['Growth_Rate'] < (-1 * std_growth))
]
lifecycle_choices = ['Introduction', 'Growth', 'Maturity', 'Decline']

# Apply classification with default 'Maturity' for edge cases
lifecycle_data['Lifecycle_Phase'] = np.select(lifecycle_conditions, lifecycle_choices, default='Maturity')

# Step 6: Summarize distribution
phase_distribution = lifecycle_data['Lifecycle_Phase'].value_counts()
phase_percentages = lifecycle_data['Lifecycle_Phase'].value_counts(normalize=True) * 100

print("\n" + "-" * 40)
print("LIFECYCLE PHASE DISTRIBUTION")
print("-" * 40)
for phase in phase_distribution.index:
    count = phase_distribution[phase]
    pct = phase_percentages[phase]
    print(f"  {phase}: {count} months ({pct:.1f}%)")

# Step 7: Identify dominant phase
dominant_phase = phase_distribution.index[0]
print(f"\nDominant Phase: {dominant_phase}")

# Interpretation based on dominant phase
interpretations = {
    'Introduction': f"{assigned_language} is in early adoption phase with slow but positive growth.",
    'Growth': f"{assigned_language} is experiencing rapid growth and gaining popularity.",
    'Maturity': f"{assigned_language} is in the maturity phase with stable global usage.",
    'Decline': f"{assigned_language} shows signs of declining popularity."
}
print(f"Interpretation: {interpretations.get(dominant_phase, 'Analysis complete.')}")

# Step 8: Display sample table
print("\n" + "-" * 40)
print("SAMPLE LIFECYCLE TABLE (First 10 rows)")
print("-" * 40)
display_cols = ['Month', 'Popularity', 'Growth_Rate', 'Moving_Avg', 'Moving_STD', 'Lifecycle_Phase']
print(lifecycle_data[display_cols].head(10).to_string())
```

Additional Exercises # 3: Life Cycle Classification of Programming Languages
Functionality: Perform a comparative analysis between two programming languages to evaluate their performance trends, variability, and dominance over 20 years. Apply NumPy and Pandas statistical operations to compute comparative metrics such as mean difference, correlation, dominance ratio, and relative performance index.

Input:

    • Dataset file: most-popular-programming-languages-2004-2024.csv (downloaded from Kaggle)

    • Columns used:

        − Month

        − Two-language column

    • Language Selection Rule:

        − First language → based on the last three digits of the student ID

        − Second language → next language alphabetically after the first

Calculations:
    
    • Load dataset using Pandas read_csv() and parse Month as datetime.

    • Select the two assigned languages for analysis.

    • Compute central tendencies for each language:
    
        − Mean (.mean()), Median (.median()), Standard Deviation (.std())
    
    • Normalize data using NumPy to align scales.

    • Calculate correlation coefficient using np.corrcoef(lang1, lang2)[0,1]
    
    • Determine dominance ratio:
    
        − Count how many months Language A > Language B
    
        − Dominance Ratio = (Count A > B) / Total Months × 100

    • Compute relative performance index (RPI)
    
    • Identify cross-over points (months when A surpasses B) using Boolean indexing
    
    • Summarize comparative statistics in a DataFrame:
        
        − Mean_A, Mean_B, Std_A, Std_B, Correlation, Dominance_Ratio, RPI

    • Interpret which language demonstrates more consistent or resilient growth.

Output:

    • Lifecycle Classification Table including: 
        
        − Month

        − Popularity (%)

        − Growth_Rate (%)
    
        − Moving_Avg

        − Moving_STD
    
        − Lifecycle_Phase

Summary Output:

    − Count and percentage per lifecycle phase
    
    − Dominant stage (e.g., “Python is currently in the maturity phase with stable global adoption.”)

```python

# ANSWER FOR ADDITIONAL EXERCISE #3: Comparative Analysis of Two Programming Languages

print("EXERCISE #3: Comparative Language Analysis")

# Step 1: Select two languages
lang1_name = assigned_language
lang2_name = second_language

print(f"\nComparing: {lang1_name} vs {lang2_name}")

# Step 2: Extract data for both languages
lang1_col = f'{lang1_name} Worldwide (%)'
lang2_col = f'{lang2_name} Worldwide (%)'

# Handle column name matching (same logic as Exercise 1)
if lang1_col not in df.columns:
    matching = [col for col in df.columns if lang1_name.lower() in col.lower()]
    if matching:
        lang1_col = matching[0]
    else:
        lang1_col = df.columns[1]

if lang2_col not in df.columns:
    matching = [col for col in df.columns if lang2_name.lower() in col.lower()]
    if matching:
        lang2_col = matching[0]
    else:
        lang2_col = df.columns[2] if len(df.columns) > 2 else df.columns[1]

# Step 3: Create comparison DataFrame
compare_df = pd.DataFrame({
    'Month': df['Month'],
    f'{lang1_name}_Pop': df[lang1_col],
    f'{lang2_name}_Pop': df[lang2_col]
})

# Step 4: Calculate central tendencies
mean_1 = compare_df[f'{lang1_name}_Pop'].mean()
mean_2 = compare_df[f'{lang2_name}_Pop'].mean()
median_1 = compare_df[f'{lang1_name}_Pop'].median()
median_2 = compare_df[f'{lang2_name}_Pop'].median()
std_1 = compare_df[f'{lang1_name}_Pop'].std()
std_2 = compare_df[f'{lang2_name}_Pop'].std()

print("\n" + "-" * 40)
print("CENTRAL TENDENCY COMPARISON")
print("-" * 40)
print(f"{'Metric':<20} {lang1_name:<15} {lang2_name:<15}")
print(f"{'Mean':<20} {mean_1:<15.2f} {mean_2:<15.2f}")
print(f"{'Median':<20} {median_1:<15.2f} {median_2:<15.2f}")
print(f"{'Std Dev':<20} {std_1:<15.2f} {std_2:<15.2f}")

# Step 5: Calculate correlation coefficient
# np.corrcoef returns a 2x2 matrix, [0,1] gives the correlation between the two variables
correlation = np.corrcoef(compare_df[f'{lang1_name}_Pop'], compare_df[f'{lang2_name}_Pop'])[0, 1]
print(f"\nCorrelation Coefficient: {correlation:.4f}")

# Step 6: Determine dominance ratio
# Count how many months Language 1 > Language 2
lang1_dominant = (compare_df[f'{lang1_name}_Pop'] > compare_df[f'{lang2_name}_Pop']).sum()
total_months = len(compare_df)
dominance_ratio = (lang1_dominant / total_months) * 100

print(f"\nDominance Analysis:")
print(f"  {lang1_name} > {lang2_name}: {lang1_dominant} months ({dominance_ratio:.1f}%)")
print(f"  {lang2_name} > {lang1_name}: {total_months - lang1_dominant} months ({100-dominance_ratio:.1f}%)")

# Step 7: Calculate Relative Performance Index (RPI)
# RPI = (Mean_A / Mean_B) × 100
rpi = (mean_1 / mean_2) * 100
print(f"\nRelative Performance Index (RPI): {rpi:.2f}")
print(f"  (>100 means {lang1_name} is generally more popular)")
print(f"  (<100 means {lang2_name} is generally more popular)")

# Step 8: Identify cross-over points
# Find months where Language 1 surpasses Language 2
compare_df['Lang1_Dominant'] = compare_df[f'{lang1_name}_Pop'] > compare_df[f'{lang2_name}_Pop']
# A cross-over happens when the boolean value changes from False to True
crossovers = compare_df[compare_df['Lang1_Dominant'] != compare_df['Lang1_Dominant'].shift(1)]

if len(crossovers) > 1:
    print(f"\nCross-over Points Found: {len(crossovers) - 1}")
    print("(Times when leadership changed between languages)")
else:
    print(f"\nNo significant cross-over points detected.")

# Step 9: Summary DataFrame
print("\n" + "-" * 40)
print("COMPARATIVE STATISTICS SUMMARY")
print("-" * 40)
summary_df = pd.DataFrame({
    'Metric': ['Mean', 'Median', 'Std Dev', 'Correlation', 'Dominance Ratio (%)', 'RPI'],
    lang1_name: [mean_1, median_1, std_1, correlation, dominance_ratio, rpi],
    lang2_name: [mean_2, median_2, std_2, correlation, 100-dominance_ratio, 100]
})
print(summary_df.to_string(index=False))

# Step 10: Interpretation
print("\n" + "-" * 40)
print("INTERPRETATION")
print("-" * 40)
if rpi > 110:
    print(f"{lang1_name} shows significantly stronger and more consistent popularity.")
elif rpi < 90:
    print(f"{lang2_name} shows significantly stronger and more consistent popularity.")
else:
    print("Both languages show comparable popularity levels with balanced competition.")

if std_1 < std_2:
    print(f"{lang1_name} demonstrates more stable/consistent popularity (lower volatility).")
else:
    print(f"{lang2_name} demonstrates more stable/consistent popularity (lower volatility).")
```
